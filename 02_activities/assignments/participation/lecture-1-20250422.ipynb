```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# making some sample data by setting a seed

np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)
```

```python
# making a basic scatterplot of data
# fig is the container, ax is the object
fig, ax = plt.subplots(figsize=(5,3)) # creating a 5 by 3 inch plot
ax.scatter(x,y)
```

```python
# making the ax object a bar graph now
fig, ax = plt.subplots(figsize=(5,3))
ax.bar(x,y) # vertical
```

```python
# can create horizontal bar graph by using barh
fig, ax = plt.subplots(figsize=(5,3)) # don't have to specify the size each time, there is a default size
ax.barh(x,y)
```

fig is the container and anything we add into the figure (such as the axes) are objects within that container. 

```python
# creating a line plot
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)
```

```python
# creating a histogram
fig, ax = plt.subplots(figsize=(5,3))
ax.hist(y) # histogram only counts frequency so only need the y-value
```

```python
# adding axis labels and titles
# important to define elements of object (axes) before we customize it
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

ax.set_title("Total growth over time") # plot title
ax.set_ylabel("Total growth") # y-axis title
ax.set_xlabel("Years since start") # x-axis title
fig.tight_layout() # this makes sure all labels are aligned to the plot
```

```python
# modifying the labels and titles (such as font)

# first we define the dictionary by setting out the properties we want
font1 = {'family': 'sans-serif', 'color': 'blue', 'size': 20}
font2 = {'family': 'monospace', 'color': 'green', 'size': 14} # note that the numbers are not in quotations
```

```python
# assigning the font dictionaries to our graph

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

ax.set_title("Total growth over time",  fontdict = font1, loc = 'left') # loc changes alignment of title
ax.set_ylabel("Total growth", fontdict = font2) 
ax.set_xlabel("Years since start", fontdict = font2) 
fig.tight_layout()
```

```python
# modifying our data points

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,
           y,
           marker = "*",
           color = "indigo") # changing our marker to an asterisk/star shape that is an indigo color
fig.show()
```

```python
# can also change linewidth and style

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,
           y,
           marker = "*",
           color = "indigo",
           linestyle = "--", # changed to dashed style
           linewidth = 2) 
fig.show()
```

```python
# don't just have to modify colours by naming them, can also use hex codes

fig, ax = plt.subplots(figsize=(5,3))

# color is the new addition here
ax.plot(x,
           y,
           marker = "*",
           color = "#7425b9",
           linestyle = "--", # changed to dashed style
           linewidth = 2) 
fig.show()
```

```python
# customizing the plot further

fig, ax = plt.subplots(figsize=(5,3))

# color is the new addition here
ax.plot(x,
           y,
           marker = "*",
           markersize = 12,
           color = "#7425b9",
           linestyle = "--", # changed to dashed style
           linewidth = 2,
           markeredgecolor = "#fa9359",
           markerfacecolor = "#000000") 
fig.show()
```

```python
# can also add grid lines
fig, ax = plt.subplots(figsize=(5,3))

# color is the new addition here
ax.plot(x,
           y,
           marker = "*",
           markersize = 12,
           color = "#7425b9",
           linestyle = "--", # changed to dashed style
           linewidth = 2,
           markeredgecolor = "#fa9359",
           markerfacecolor = "#000000") 

ax.grid() # this generates x and y grind lines in a default gray colour
```

```python
# can change the colour of the grid lines as well
fig, ax = plt.subplots(figsize=(5,3))

# color is the new addition here
ax.plot(x,
           y,
           marker = "*",
           markersize = 12,
           color = "#7425b9",
           linestyle = "--", # changed to dashed style
           linewidth = 2,
           markeredgecolor = "#fa9359",
           markerfacecolor = "#000000") 

# changing colour of grid lines
ax.grid(color = "#7425b9", # changed color
        linestyle = "--", # changed to dashed style
        linewidth = 2) # changed width

# can also specify which grid line (x or y) we want to modify 
ax.grid(axis = "y", 
        color = "#7425b9", # changed color
        linestyle = "--", # changed to dashed style
        linewidth = 2) # changed width
```

```python
# can also specify which grid line (x or y) we want to modify 
fig, ax = plt.subplots(figsize=(5,3))

# color is the new addition here
ax.plot(x,
           y,
           marker = "*",
           markersize = 12,
           color = "#7425b9",
           linestyle = "--", # changed to dashed style
           linewidth = 2,
           markeredgecolor = "#fa9359",
           markerfacecolor = "#000000") 

# only modifying y-axis of grid line
ax.grid(axis = "y", 
        color = "#7425b9", # changed color
        linestyle = "--", # changed to dashed style
        linewidth = 2) # changed width
```

```python
fig, ax = plt.subplots(figsize=(5,3))

# color is the new addition here
ax.plot(x,
           y,
           marker = "*",
           markersize = 12,
           color = "#7425b9",
           linestyle = "--", # changed to dashed style
           linewidth = 2,
           markeredgecolor = "#fa9359",
           markerfacecolor = "#000000") 

# change linestyle
ax.grid(axis = "y", 
        color = "blue", # changed color
        linestyle = "-.", # changed to dashed style
        linewidth = 2) # changed width
```

```python
# Matplotlib Activity
import seaborn as sns

# load dataset
df = sns.load_dataset('iris')

# create simple density plot
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length)
plt.show()

# add colur (cmap) and smoothing (bw_adjust) to make it easier to read 
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length, cmap= "Blues", fill = True, bw_adjust=0.5)

# define font
font = {'family': 'sans-serif', 'color': 'black', 'size': 12}

# add x and y axis titles
plt.xlabel("Sepal Width", fontdict = font) 
plt.ylabel("Sepal Length", fontdict = font) 

plt.show()
```

By adjusting the colour and bandwith of the kernel density estimate, we can better understand the relationship between sepal length and width.
