```python
# load libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

# sample data
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

```python
# lineplot with both y variables on the same axes
fig, ax = plt.subplots(figsize=(5,3))

ax.plot(x,y1)
ax.plot(x,y2)
```

```python
# add a legend
fig, ax = plt.subplots(figsize=(5,3))

ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc="lower right") # location of the legend
```

```python
# modify a legend
fig, ax = plt.subplots(figsize=(5,3))

ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc="lower right",
          frameon = True, # added frame around legend
          fontsize = 12, 
          ncol = 2, # appears as one line since each person in a separate column instead of one column/or list
          shadow = True) # added shadow
```

```python
# move legend outside of the plot area
fig, ax = plt.subplots(figsize=(5,3))

# bbox_to_anchor specifies the coordinates of where you want the legend to go
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc="upper left", # using this as the anchor point
          bbox_to_anchor = (1,0.60)) # these coordinates are based on the anchor point above
# 1, 1 represents 100% on horizontal axis and vertical axis, so it wil move it to the top right
# changed to 0.60 so that legend can be in the middle
```

```python
# adding text annotation
fig, ax = plt.subplots(figsize=(5,3))

ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc="lower right")

ax.text(10,95, "This value is important!") # this is referring to the point at the position 10,95 (the dot at the beginning of the text)
```

```python
# modifying text annotation
fig, ax = plt.subplots(figsize=(5,3))

ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc="lower right")

ax.text(10,95, "This value is important!",
        ha = "center", # ha is horizontal alignment, so the text is aligned center to the point that we are talking about
        color = "red",
        size = 20) 
```

```python
# positioning text annotation
fig, ax = plt.subplots(figsize=(5,3))
ax.axis([0,10,0,10]) # defining x and y ranges

ax.text(1, 5, ". Data:(1, 5)", transform = ax.transData) # the number specific the coordinates of the position
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)", transform=ax.transAxes)
ax.text(0.2, 0.2, ". Figure:(0.2,0.2)", transform= fig.transFigure)
```

```python
# annotating with arrows
fig, ax = plt.subplots(figsize=(5,3))

ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc="lower right")

ax.annotate("This is important!", xy=(10,95), # xy location of where we want the arrow to go
            xytext=(20,94), # xy location of where we want the text to go
            arrowprops = dict(facecolor = "black", 
                              shrink = 0.05)) # this specifices the amount of gap between arrows top and stem, and the text 
```

```python
# modyfing arrows
fig, ax = plt.subplots(figsize=(5,3))

ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc="lower right")

ax.annotate("This is important!", xy=(10,95), 
            xytext=(20,94), 
            arrowprops = dict(arrowstyle = "wedge",
                              color = "hotpink")) # if you use facecolor, there's a black border around the arrow with pink inside. using just color means no border
```

```python
# removing tick marks and labels
fig, ax = plt.subplots(figsize=(5,3))

ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc="lower right")

ax.yaxis.set_major_locator(plt.NullLocator()) # for y-axis, notches and labels are both gone
ax.xaxis.set_major_formatter(plt.NullFormatter()) # for x-axis, still have tick marks/notches, but no labels

```
